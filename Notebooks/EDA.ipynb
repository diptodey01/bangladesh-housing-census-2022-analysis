## Figuring out the total number of numeric columns

```python

numeric_columns = df.select_dtypes(include=["number"]).columns

print("Number of numeric columns:", len(numeric_columns))

print(df[numeric_columns].describe().T)
```

## Checking if there are any negative values. 

```python

numeric_columns = df.select_dtypes(include=["number"]).columns

negative_values = (df[numeric_columns] < 0).sum()

print(negative_values[negative_values > 0])
```

## Checking if there are any infinite values. 

```python

infinite_values = np.isinf(df[numeric_columns]).sum()

print(infinite_values[infinite_values > 0])
```

## Checking column with percentage and their validity

```python
percentage_columns = [ column for column in df.columns if "%" in column or "percent" in column.lower() or "percentage" in column.lower()] 

print("Number of percentage columns:", len(percentage_columns))

for column in percentage_columns:
 print(column)
```

```python
invalid_percentage_columns = []

for column in percentage_columns:
    invalid = df[(df[column] < 0) | (df[column] > 100)]

    if not invalid.empty:
        invalid_percentage_columns.append(column)
        print(f"\nInvalid values in: {column}")
        print(invalid[["Division", "District", column]])

if not invalid_percentage_columns:
    print("All percentage columns contain valid values between 0% and 100%.")
```

Here are 158 columns with percentages that contain valid values between 0% and 100%."
