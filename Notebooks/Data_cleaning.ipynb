# Bangladesh Housing & Population Census - Data cleaning

## 1. Setup & Data Loading

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("housing_census.csv")
```

## Standardizing column names

```python
df.columns= (df.columns.str.strip().str.lower().str.replace(r"[^\w\s]","",regex= True).str.replace(r"\s+","_",regex=True))
```

```python

df.columns = df.columns.str.replace("'", "").str.strip().str.lower()
```

```python
df.head()
```

```python
df.tail()
```

## Fixing Categorical & identifier data types

```python
df["division_geocode"] = df["division_geocode"].astype(str)
df["district_geocode"] = df["district_geocode"].astype(str)
```

```python
df.head()
```

```python
df.describe()
```

```python
for i, column in enumerate(df.columns, start=1):
    print(f"{i}.{column}")
```

## Replacing literal 's' (or similar typos) with NaN

```python

df = df.replace(['s', 'S', ' ', ''], np.nan)
```

## Updated Dataframe

```python
df_clean=df.copy()


housing_census_clean=df_clean.to_csv("housing_census_cleaned.csv", index=False)


```

```python
df_clean.head()
```

## Verifying total population number for all district. 

```python

total_col = "type_of_dwelling_sex_dist_population_total_"
male_col = "type_of_dwelling_sex_dist_population_male_"
female_col = "type_of_dwelling_sex_dist_population_female_"
hijra_col = "type_of_dwelling_sex_dist_population_hijra_"


male_series = pd.to_numeric(df_clean[male_col], errors="coerce").fillna(0)
female_series = pd.to_numeric(df_clean[female_col], errors="coerce").fillna(0)
hijra_series = pd.to_numeric(df_clean[hijra_col], errors="coerce").fillna(0)
total_series = pd.to_numeric(df_clean[total_col], errors="coerce")


calculated = male_series + female_series + hijra_series


print(np.isclose(total_series, calculated, equal_nan=True).all())
```
